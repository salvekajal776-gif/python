```python
# Univariate Data
```

```python
# Univariate means:
# “One variable” (uni = one)

# You are analyzing only a single feature/column of data.
```

```python
# Examples

# Student marks
# Person height
# Product price

# You are NOT comparing — just understanding one variable.
```

```python
# What do we study in Univariate data?

# Distribution (shape)
# Central tendency (mean, median)
# Spread (variance, range)
# Outliers
```

```python
# KDE plot (Kernel Density Estimate) 
# shows the smooth probability distribution of data instead of bars like a histogram.
# It helps you see where values are concentrated and how the data is shaped (peaks, spread, skewness).
# Think of it as a “smoothed histogram” that makes patterns easier to understand visually.
```

```python
# Univariate KDE
```

```python
data = sns.load_dataset("iris")
data.head(5)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset("iris")

# Univariate: only one variable
sns.kdeplot(data=data, x="sepal_length", fill=True)

plt.title("Univariate Distribution")
plt.show()
```

```python
# Bivariate Data
```

```python
# Bivariate means:

# “Two variables” (bi = two)
# You analyze relationship between two variables
```

```python
# Examples

# Height vs Weight
# Study hours vs Marks
# Advertising cost vs Sales

# Now you're asking:
# “Does one variable affect the other?”
```

```python
# What do we study in Bivariate data

# Relationship (correlation)
# Trend (positive/negative)
# Patterns or clusters
```

```python
# Bivariate KDE
```

```python
data = sns.load_dataset("iris")

sns.kdeplot(
    data=data,
    x="sepal_length",
    y="sepal_width",
    fill=True
)

plt.title("Bivariate Distribution")
plt.show()
```

```python
# Joint Plot
```

```python
# A Joint Plot is a visualization that combines:

# Bivariate analysis (relationship between two variables)
# Univariate analysis (distribution of each variable)
```

```python
# A Scatter Joint Plot is a type of joint plot where:
# The center plot = scatter plot
# The margins = histograms or KDE or etc

# It shows actual data points instead of smooth density.
```

```python
data = sns.load_dataset("iris")

sns.jointplot(
    data=data,
    x="sepal_length",
    y="sepal_width",
    kind="scatter"
)

plt.show()
```

```python
# Multidimensional exploration

# Multidimensional exploration means analyzing a dataset that has many variables (features) 
# at the same time, instead of looking at them one by one.
```

```python
# When data has multiple dimensions—like height, weight, age, income, score, etc.—the goal is to understand:
# How each variable behaves individually
# How variables relate to each other
# Whether patterns, clusters, or trends exist across combinations of variables
```

```python
# Pair Plot (pairplot)
# Shows relationships between all numeric columns.
```

```python
data = sns.load_dataset("tips")

sns.pairplot(data)
plt.show()
```

```python
# Regression visualization
```

```python
# Regression visualization using lmplot (from Seaborn) helps you:
# Plot raw data points (scatter plot)
# Fit a linear regression line automatically
# Optionally split plots into multiple subplots using a FacetGrid (categories)
```

```python
# Load sample dataset
df = sns.load_dataset("tips")

# Linear regression plot
sns.lmplot(
    data=df,
    x="total_bill",
    y="tip"
)

plt.show()
```

```python
# This shows:
# Scatter points (data)
# Best-fit regression line
# Confidence interval (shaded area)
```

```python
# lmplot with Category (FacetGrid effect)
# You can split the plot by categories using hue:
```

```python
df = sns.load_dataset("tips")

sns.lmplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="sex"
)

plt.show()
```

```python
# Now you get:
# Separate regression lines for male and female
# Color-coded groups
```
